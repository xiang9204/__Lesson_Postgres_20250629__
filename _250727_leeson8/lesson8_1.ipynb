```python
#建立 tuple，其資料建立後，無法再被修改，為暫時儲存用
student1 = (87, 76, 65, 78, 92) #sequence (串列資料 or 序列資料)
student1
type(student1)
```

```python
#建立 list
student2 = [87, 76, 65, 78, 92] #sequence (串列資料)
student2
type(student2)

#使用 subscript 取出元素值，索引值從 0 開始
student2[0]
student2[1]
student2[2]
student2[3]
student2[4]
#student2[5] # 超出索引會出錯
```

```python
#建立 dict (dictionary)
student3 = {'chinese':67,'english':75,'math':89,'history':87,'discovery':92} #mapping(映射資料)
type(student3)
#使用 subscript 取出值
print(student3['math'])
print(student3['history'])
```
