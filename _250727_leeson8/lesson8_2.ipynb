```python
#請建立一個範本 list
my_list = [1, 2, 3, 4, 5] #sequence (串列資料)
print("my_list =", my_list)
#並且建立常用的實體方法
my_list.append(6)  # 在 list 的尾端加入元素
print("my_list after append(6) =", my_list)
my_list.insert(0, 0)  # 在 list 的開頭加入元素
print("my_list after insert(0, 0) =", my_list)
my_list.remove(3)  # 移除 list 中的元素
print("my_list after remove(3) =", my_list)
# 在 list 中的元素排序
my_list.sort()  # 預設是從小到大排序
print("my_list after sort() =", my_list)
my_list.reverse()  # 將 list 中的元素反轉
print("my_list after reverse() =", my_list)
my_list.clear()  # 清空 list 中的所有元素
print("my_list after clear() =", my_list)
```

```python
# 一個 list, 使用 for loop, 將 list 中的每個元素乘以2
my_list = [1, 2, 3, 4, 5]
for i in my_list:
    print(i * 2)  # 印出每個元素乘以2的結果

for letter in "Apple":
    print(letter)  # 印出每個字母
```

```python
# 請建立國家 code 與國家名稱的 dictionary
country_codes = {
    "TW": "Taiwan",
    "US": "United States",
    "JP": "Japan",
    "CN": "China",
    "KR": "South Korea"
}
print("country_codes =", country_codes)
# 並且建立常用的實體方法
country_codes["FR"] = "France"  # 新增一個國家
print("country_codes after adding FR =", country_codes)
del country_codes["CN"]  # 刪除一個國家 
print("country_codes after deleting CN =", country_codes)
print("Is 'TW' in country_codes?", "TW" in country_codes)  # 檢查 'TW' 是否在 dictionary 中
print("Keys in country_codes:", country_codes.keys())  # 印出所有的鍵
print("Values in country_codes:", country_codes.values())  # 印出所有的值
print("Items in country_codes:", country_codes.items())  # 印出所有的鍵值
print("Country code for Taiwan:", country_codes.get("TW"))  # 取得台灣的國家名稱
print("Country code for India:", country_codes.get("IN", "Not Found"))  # 取得印度的國家名稱，如果不存在則回傳 "Not Found"
```

```python
for code in country_codes.keys():  # 遍歷所有的國家代碼
    print(code, ":", country_codes[code])  # 印出國家代碼與對應的國家名稱
```

```python
for country in country_codes.values():  # 遍歷所有的國家名稱
    print(country)  # 印出國家名稱
```

```python
for item in country_codes.items():  # 遍歷所有的鍵值對
    print(item)  # 印出鍵值對
```

```python
for key, value in country_codes.items():  # 遍歷所有的鍵值對
    print(key, ":", value)  # 印出鍵與值
```
