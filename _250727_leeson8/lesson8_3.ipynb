### 自訂 function (函式)

```python
def add(a, b):
    c = a + b
    return c

result = add(3,4)  # 呼叫函式，傳入參數1和2，返回3
print("result =",result)  # 印出結果3
```

```python
#定義一個函式，名稱為sayHello，接受一個參數name
def sayHello(name):
    print("Hello!", name)

sayHello("Xiang")  # 呼叫函式，傳入參數"Xiang"，印出"Hello Xiang"
```

```python
#一個函式，輸入 name 參數，輸出 Hello, 名字
#以字串插補方式，產生組合後的新字串
def greet(name):
    print(f"Hello,{name}")

# 呼叫 greet 函式，傳入參數 "Xiang"
greet("Xiang")  # 印出 "Hello, Xiang"
```

```python
#建立一個 function, 傳入一個數字陣列，回傳總和
def sum_array(arr):
    total = 0
    for num in arr:
        total += num
    return total

#建立一個數字陣列
numbers = [2, 4, 6, 8, 10]  # 數字陣列
#呼叫 sum_array 函式，傳入 numbers 陣列，並印出
print("Sum of numbers:", sum_array(numbers))  # 印出數字陣列的總和
```

```python
#取餘數
def remainder(a, b):
    return a % b
#印出餘數
print("Remainder of 10 divided by 3:", remainder(10, 3))  # 印出10除以3的餘數   
```
